### Q15. Indexing and slicing to build X and y
Create a 2-D array with 6 rows and 3 columns, where the last column is a target label (0 or 1) and the first two columns are features (make up your own numbers).
1. Print the first 3 rows only.
2. Print only the first column, for all rows.
3. Split the array into `X` (all columns except the last) and `y` (only the last column), and print both.

```python
# Write your code here
import numpy as np
arr = np.array([[10, 20, 0],[15, 25, 1],[30, 40, 1],[12, 18, 0],[35, 45, 1],[22, 28, 0]])
# 1. Print the first 3 rows
print("First 3 rows:")
print(arr[:3])
# 2. Print the first column
print("First column:")
print(arr[:, 0])
# 3. Split into X and y
X = arr[:, :-1]
y = arr[:, -1]
print("X:")
print(X)
print("y:")
print(y)
```

### Q16. Vectorized operations and boolean masking
Create a 1-D NumPy array of 8 exam scores (0-100) of your choice. Without using any loop:
1. Add 5 bonus marks to every score and print the result.
2. Print a boolean array showing which scores are 50 or above.
3. Print only the scores that are 50 or above using boolean masking.

```python
# Write your code here
import numpy as np
scores = np.array([45, 67, 82, 39, 55, 91, 48, 73])
# 1. Add 5 bonus marks
bonus_scores = scores + 5
print("Scores after bonus:")
print(bonus_scores)
# 2. Boolean array for scores 50 or above
mask = scores >= 50
print("Boolean array:")
print(mask)
# 3. Print scores 50 or above
print("Scores 50 or above:")
print(scores[mask])
```
